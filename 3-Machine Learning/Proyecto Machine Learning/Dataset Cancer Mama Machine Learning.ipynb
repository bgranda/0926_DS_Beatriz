This dataset of breast cancer patients was obtained from the 2017 November update of the SEER Program of the NCI, which provides information on population-based cancer statistics. The dataset involved female patients with infiltrating duct and lobular carcinoma breast cancer (SEER primary cites recode NOS histology codes 8522/3) diagnosed in 2006-2010. Patients with unknown tumour size, examined regional LNs, positive regional LNs, and patients whose survival months were less than 1 month were excluded; thus, 4024 patients were ultimately included.

El sistema TNM es el estándar internacional utilizado para describir la extensión anatómica del cáncer. En cáncer de mama, las letras significan:

T (Tumor): tamaño y extensión del tumor primario.
N (Nodes): afectación de ganglios linfáticos regionales.
M (Metastasis): presencia o ausencia de metástasis a distancia.

T: Tumor primario
Categoría	Definición
Tis	Carcinoma in situ (no invasivo)
T1	Tumor ≤ 20 mm
T1mi	≤ 1 mm (microinvasión)
T1a	> 1 mm y ≤ 5 mm
T1b	> 5 mm y ≤ 10 mm
T1c	> 10 mm y ≤ 20 mm
T2	> 20 mm y ≤ 50 mm
T3	> 50 mm
T4	Cualquier tamaño con invasión de pared torácica o piel
T4a	Invasión de pared torácica
T4b	Afectación cutánea (úlceras, nódulos, edema)
T4c	T4a + T4b
T4d	Cáncer inflamatorio de mama

Estadio == 6th Stage
Resumen simplificado
Estadio	Situación general
0	Carcinoma in situ
I	Tumor pequeño localizado
II	Tumor más grande o pocos ganglios afectados
III	Enfermedad localmente avanzada
IV	Metástasis a distancia

Relación con el pronóstico

Peor supervivencia.

2. Grado tumoral (Grade)

El grado es una medida de agresividad basada en la diferenciación celular.

Grade	Significado
1	Bien diferenciado
2	Moderadamente diferenciado
3	Poco diferenciado
4	Indiferenciado o anaplásico

Por tanto, en muchos datasets:

Plain Text
Grade 1 ≈ Well differentiated
Grade 2 ≈ Moderately differentiated
Grade 3 ≈ Poorly differentiated
Grade 4 ≈ Undifferentiated
Diferencia entre Grade y Differentiation

En realidad describen casi la misma característica biológica.

Por eso, durante el análisis del dataset conviene estudiar si ambas variables son redundantes antes de incluirlas simultáneamente en el modelo.

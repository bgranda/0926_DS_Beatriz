Dataset Highlights
280,985 patient records
38 standardized features
Zero missing values
Binary and multi-class target labels
Integrated from three healthcare datasets
Suitable for machine learning, AI, and healthcare analytics
Feature Categories
The dataset includes:

Demographic information (age, gender, age group)
BMI and anthropometric measurements
Lifestyle factors (smoking, physical activity, stress, sleep, alcohol, salt, and sugar intake)
Clinical measurements (glucose, HbA1c, cholesterol, triglycerides, HDL, LDL, blood pressure, heart rate, CRP, and homocysteine)
Disease indicators and classification targets
Column Details
composite_key: Unique patient profile identifier.
age_level: Age group (Young, Middle, Mature, Senior).
gender: Patient gender.
bmi_level: BMI category based on WHO standards.
smoking: Smoking status.
diabetes: Diabetes indicator.
age: Patient age in years.
age_normalized: Normalized age (0–1).
bmi: Body Mass Index (kg/m²).
hypertension: Hypertension indicator.
heart_disease: Heart disease indicator.
HbA1c_level: Long-term blood glucose indicator.
glucose: Blood glucose level.
cholesterol: Total cholesterol level.
sleep_hours: Average daily sleep duration.
triglycerides: Triglyceride level.
physical_activity: Physical activity level.
family_history: Family history of chronic disease.
stress_level: Stress level.
low_hdl_cholesterol: Low HDL cholesterol indicator.
high_ldl_cholesterol: High LDL cholesterol indicator.
blood_pressure: Blood pressure measurement.
high_blood_pressure: High blood pressure indicator.
sugar_consumption: Daily sugar intake level.
crp_level: C-Reactive Protein (CRP) level.
homocysteine_level: Homocysteine level.
systolic_bp: Systolic blood pressure.
diastolic_bp: Diastolic blood pressure.
alcohol_intake: Alcohol consumption level.
salt_intake: Dietary salt intake level.
heart_rate: Resting heart rate.
hdl: HDL cholesterol level.
ldl: LDL cholesterol level.
education_level: Education level.
employment_status: Employment status.
source_dataset: Original source dataset.
disease_flags: Disease combination code.
sublabel: Multi-class disease label (8 classes).
label: Binary target label (Normal or Abnormal).

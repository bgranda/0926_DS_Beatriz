## Building a regression model

The data have already been split to predictor (*data*) and response (*target*) variables. Given this information, we'll apply what we have previously learned about linear regression and apply the algorithm to the diabetes dataset.

#### In the cell below, import the `linear_model` class from `sklearn`. 

```python
from sklearn.linear_model import LinearRegression
```

#### Create a new instance of the linear regression model and assign the new instance to the variable `diabetes_model`.

```python
diabetes_model = LinearRegression()
diabetes_model
```

#### Next, let's split the training and test data.

Define `diabetes_data_train`, `diabetes_target_train`, `diabetes_data_test`, and `diabetes_target_test`. Use the last 20 records for the test data and the rest for the training data.

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns='target')
y = df['target']

diabetes_data_train = X.iloc[:-20]
diabetes_target_train = y.iloc[:-20]
diabetes_data_test = X.iloc[-20:]
diabetes_target_test = y.iloc[-20:]
```

```python
print("Total features shape:", X.shape)
print("Train features shape:", diabetes_data_train.shape)
print("Train target shape:", diabetes_target_train.shape)
print("Test features shape:", diabetes_data_test.shape)
print("Test target shape:", diabetes_target_test.shape)
```

Perform a brief exploratory analysis to investigate the conditions necessary for linear regression. This analysis will help you understand the distribution of the target variable and its correlations with other features.

Fit the training data and target to `diabetes_model`. Print the *intercept* and *coefficients* of the model.

```python
diabetes_model.fit(diabetes_data_train, diabetes_target_train)
```

```python
diabetes_model.coef_
```

```python
coeff_df = pd.DataFrame(diabetes_model.coef_,
                        X.columns,
                        columns=['Coefficient'])
coeff_df
```

#### Iterate with a new model to assess if any improvements can be made compared to the previous one. Consider reducing the number of variables or using `StandardScaler` to enhance model performance.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(diabetes_data_train)
X_test_scaled = scaler.transform(diabetes_data_test)

modelo = LinearRegression()
modelo.fit(X_train_scaled, diabetes_target_train)

y_pred2 = modelo.predict(X_test_scaled)

print("R²:", r2_score(diabetes_target_test, y_pred2))
```

```python
prediccion2 = diabetes_model.predict(X_test_scaled)
prediccion2
```

```python
mae = metrics.mean_absolute_error(X_test_scaled, prediccion2)
mse = metrics.mean_squared_error(diabetes_target_test, prediccion2)
rmse = metrics.root_mean_squared_error(diabetes_target_test, prediccion2)
mape = mean_absolute_percentage_error(diabetes_target_test, y_pred2)

print(f"mean absolute error", mae)
print(f"mean squared error", mse)
print(f"root mean squared error", rmse)
print(f"mean absolute percentage error", mape)
```

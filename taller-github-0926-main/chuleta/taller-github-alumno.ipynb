# Bloque 0 — Qué es Git, qué es GitHub

## Git no es GitHub (y los dos te hacen falta)

- **Git** es un programa que vive en TU ordenador. Guarda "fotos" (**commits**) de tu proyecto
  cada vez que tú se lo pides, y te deja viajar en el tiempo entre esas fotos.
  Funciona **sin internet**.
- **GitHub** es una web (un servidor) donde subes tus repositorios Git para **compartirlos y
  colaborar**: añade Pull Requests, revisiones, issues, protección de ramas...
- **¿Se puede usar GitHub sin Git instalado?** A medias: la web deja editar archivos sueltos y
  poco más. Para trabajar de verdad necesitas Git en tu ordenador. Piensa en un iceberg:
  la web es la punta; el trabajo real (historial, ramas, merges) ocurre debajo.
- Existen otras webs parecidas (GitLab, Bitbucket). Hoy usamos GitHub porque es la más usada.

**Analogía del día:** *Git es el juego de mesa; GitHub es el servidor donde guardas la partida
y juegas con otros.*

## El viaje de tus cambios: las 4 zonas

```
 TUS ARCHIVOS        EL CARRITO         TU HISTORIAL          GITHUB
 (working dir) --git add--> (staging) --git commit--> (repo local .git) --git push--> (origin)
       ^                                                                   |
       +----------------------- git pull (trae y mezcla) ------------------------+
```

1. **Working directory**: tus archivos, los que editas con VS Code.
2. **Staging area** ("el carrito"): con `git add` metes solo lo que quieres en la próxima foto.
3. **Repositorio local** (carpeta oculta `.git`): con `git commit` haces la foto. Cada foto
   tiene un **hash** (código único, p. ej. `a1b2c3d`), autor, fecha y mensaje.
4. **Remoto** (GitHub): con `git push` subes tus fotos; con `git pull` traes las de los demás.

**El ciclo que repetirás toda tu vida: `add` → `commit` → `push`.** Y al empezar a trabajar:
**`pull`**.

### La convención de los mensajes de commit

Formato: `tipo/qué_hice`. La tabla completa:

| Tipo | Cuándo se usa |
|---|---|
| `add/` | Creas un archivo, carpeta o funcionalidad nueva |
| `fix/` | Corriges un error o bug |
| `update/` | Mejoras algo que ya existía sin cambiar su esencia |
| `delete/` | Quitas código, archivos o funcionalidades |
| `refactor/` | Reestructuras el código sin cambiar lo que hace |
| `docs/` | Documentación (README, comentarios largos) |
| `style/` | Formato: espacios, indentación (nada de lógica) |
| `test/` | Añades o modificas tests |

**Errores típicos (evítalos desde el día 1):**

- ❌ `fix/bug` → ¿qué bug? ❌ `update` → ¿qué? ❌ `add/nueva_funcionalidad` → ¿cuál?
- ❌ `fix/fix_bug` (palabra repetida) ❌ `carrusel` (sin tipo ni acción)
- ✅ `fix/error_calculo_media` · ✅ `update/grafico_ventas_colores` · ✅ `add/limpieza_csv`

Regla: el mensaje debe entenderse **dentro de 6 meses**, sin abrir el código.

`````python
# Trabaja de verdad: escribe contenido en cada archivo (>> añade sin borrar)
echo "condensador de fluzo" >> tools.txt
echo "rayo" >> weather.txt

# Un archivo al carrito -> una foto. Commits pequeños, un tema por commit.
git add tools.txt
git commit -m "add/condensador"

git add weather.txt
git commit -m "add/rayo"

# Tu historial ya tiene 3 fotos
git log --oneline
`````

`````python
# MINITEST: edita tools.txt otra vez y mira qué te cuentan estos dos comandos
echo "plutonio" >> tools.txt

# git status = QUÉ archivos han cambiado (el espejo)
git status

# git diff = QUÉ líneas han cambiado exactamente (rojo quita / verde pone)
git diff

# Cuando termines de mirar, guarda el cambio:
git add tools.txt
git commit -m "add/plutonio"
`````

## 1.4 A jugar con los comandos: add, restore, rm, reset

### Las variantes de `add` (la pregunta que todos hacéis)

| Comando | Qué mete en el carrito |
|---|---|
| `git add archivo` | Solo ese archivo |
| `git add .` | Todo lo cambiado en la carpeta actual y subcarpetas (lo normal) |
| `git add -A` | Todo lo cambiado en el repo entero, estés donde estés |
| `git add *` | Ojo: lo expande el shell; NO incluye archivos ocultos (los que empiezan por `.`) |
| `git reset archivo` | Saca ESE archivo del carrito (no borra tus cambios) |

`````python
# Vamos a provocar errores a propósito para aprender a deshacerlos

# 1) Escribe algo de lo que te vas a arrepentir
echo "esto es un error" >> tools.txt

# Mira el desastre línea a línea
git diff

# 2) git restore: devuelve el archivo a la última foto guardada (cambio PERDIDO, que es lo que querías)
git restore tools.txt

# Comprueba que el archivo volvió a su estado anterior
cat tools.txt
`````

`````python
# 3) Meter y sacar del carrito sin perder nada
echo "otro cambio" >> tools.txt

# Al carrito
git add tools.txt
git status

# Lo SACO del carrito (el cambio sigue en el archivo, solo deja de estar "preparado")
git restore --staged tools.txt
git status
`````

`````python
# 4) git rm --cached: Git deja de VIGILAR un archivo, pero NO lo borra de tu disco
#    (utilísimo cuando te has colado un archivo que debía ir en .gitignore)
git rm --cached tools.txt

# Aparece como "deleted" en staged: es normal, es solo el seguimiento, el archivo sigue ahí
git status
cat tools.txt

# Confirma el cambio de seguimiento
git commit -m "delete/tools_del_seguimiento"

# (si quieres que Git vuelva a vigilarlo: git add tools.txt)
`````

### `git reset`: deshacer el último commit (sin pushear)

`git reset HEAD~1` borra la última FOTO pero **conserva tus cambios** en la carpeta.

- `HEAD` = "donde estoy ahora". `HEAD~1` = una foto antes. También valen `HEAD^` (= HEAD~1),
  `HEAD~2` (dos fotos antes) o un hash directo: `git reset a1b2c3d`.
- Modos: `--soft` (deshace el commit, deja los cambios en el carrito), `--mixed` (el que viene
  por defecto: cambios sueltos en la carpeta) y `--hard` (borra commit **Y** cambios).
- ⚠️ `--hard` pierde trabajo de verdad. Hoy no lo usamos.
- **Regla de oro:** reset solo para commits que AÚN NO has subido (sin push).

`````python
# Mira el historial antes
git log --oneline

# Borra la última foto (los cambios vuelven a estar sin commitear)
git reset HEAD~1

# Una foto menos...
git log --oneline

# ...pero tus cambios siguen vivos en la carpeta
git status

# Vuelve a guardarlos para dejarlo todo limpio
git add tools.txt
git commit -m "add/condensador_segunda_oportunidad"
`````

## 1.5 La máquina del tiempo

Con el **hash** de cualquier foto puedes ir a visitarla. Estarás en modo "solo mirar"
(Git lo llama *detached HEAD*): no pasa nada, no rompes nada, y se vuelve con `git switch -`.

`````python
# Lista el historial y copia el hash corto (p. ej. a1b2c3d) de tu PRIMER commit
git log --oneline

# Viaja a esa foto (sustituye el hash por el tuyo)
git checkout a1b2c3d

# Tus archivos son los de aquella época: tools.txt está más vacío
cat tools.txt
ls

# Vuelve al presente ("-" = donde estaba antes)
git switch -

# Todo vuelve a estar como estaba
cat tools.txt
`````

## 1.6 Ramas: mundos paralelos

Una **rama** es una línea de trabajo paralela: se crea en un segundo, es gratis, y te deja
probar cosas SIN tocar la versión buena.

**Regla del bootcamp: en `main` no se trabaja.** `main` es el escenario (lo que se enseña);
la rama es el backstage (donde se ensaya).

`HEAD` es el puntero de "dónde estoy": al cambiar de rama, HEAD se mueve... **y tus archivos
cambian con él**. Vas a verlo con tus ojos en el ejercicio del Patronus.

`````python
# Crea la carpeta del ejercicio desde la carpeta del taller
cd ~/taller-github
mkdir patronus && cd patronus

# Repo nuevo y archivo inicial
git init
echo "PATRONUS: (por decidir)" > patronus.txt
git add patronus.txt
git commit -m "add/archivo_patronus_inicial"
`````

`````python
# Crea la rama "harry" Y muévete a ella (-c = crear)
git switch -c harry

# Cambia el archivo en ESTA rama
echo "PATRONUS: ciervo" > patronus.txt
code patronus.txt

# Pega esto y guarda:
""" 
/|       |\
`__\\       //__'
   ||      ||
 \__`\     |'__/
   `_\\   //_'
   _.,:---;,._
   \_:     :_/
     |@. .@|
     |     |
     ,\.-./ \
     ;;`-'   `---__________-----.-.
     ;;;                         \_\
     ';;;                         |
      ;    |                      ;
       \   \     \        |      /
        \_, \    /        \     |\
          |';|  |,,,,,,,,/ \    \ \_
          |  |  |           \   /   |
          \  \  |           |  / \  |
           | || |           | |   | |
           | || |           | |   | |
           | || |           | |   | |
           |_||_|           |_|   |_|
          /_//_/           /_/   /_/
""";

git add patronus.txt
git commit -m "add/patronus_ciervo"

# En la rama harry: ciervo
cat patronus.txt

# Vuelve a main...
git switch main

# ...y en main el archivo es OTRO: "(por decidir)". Los archivos CAMBIAN al cambiar de rama.
cat patronus.txt
`````

`````python
# Segunda rama (nace de main, que aún no tiene el ciervo)
git switch -c snape
echo "PATRONUS: cierva" > patronus.txt

code patronus.txt

# Pega esto y guarda:

"""
SNAPE'S PATRONUS
                   .     _,
                   |`\__/ /
                   \  . .(
                    | __T|
                   /   |
      _.---======='    |
     //               {}
    `|      ,   ,     {}
     \      /___;    ,'
      ) ,-;`    `\  //
     | / (        ;||
     ||`\\        |||
     ||  \\       |||
     )\   )\      )||
     `"   `"      `""
""";

git add patronus.txt
git commit -m "add/patronus_cierva"

# Vuelve a main y fusiona (merge) la rama harry:
git switch main
git merge harry
# -> "Fast-forward": main no se había movido, Git solo desliza la etiqueta. Sin drama.

cat patronus.txt
# ciervo

# Ahora fusiona snape: las dos ramas tocaron la MISMA línea...
git merge snape
# -> CONFLICT (content): Merge conflict in patronus.txt
# -> Automatic merge failed; fix the conflict and then commit the result.
`````

### Tu primer conflicto de merge

**No has roto nada.** Git está diciendo: "las dos ramas escribisteis en la misma línea;
NO decido yo, decide el humano". Si abres `patronus.txt` verás los **marcadores**:

```
<<<<<<< HEAD
PATRONUS: ciervo
=======
PATRONUS: cierva
>>>>>>> snape
```

Traducción: entre `<<<<<<< HEAD` y `=======` está lo que hay en la rama donde estás (main);
entre `=======` y `>>>>>>>` lo que trae la otra rama (snape).

**EL RITUAL DEL CONFLICTO (tatúatelo):**

1. Abre el archivo y **decide**: te quedas con una versión, con la otra, o escribes una mezcla.
2. **Borra las 3 líneas de marcadores** (`<<<<<<<`, `=======`, `>>>>>>>`).
3. `git add <archivo>`
4. `git commit` (Git ya te sugiere el mensaje de merge)
5. Comprueba: `git status` y `git log --oneline`

**Botón del pánico legítimo:** si quieres rendirte a mitad del merge, `git merge --abort`
y todo vuelve a como estaba.

> Si el archivo en conflicto fuera un `.ipynb` (notebook), NO se editan los marcadores a mano
> (el notebook es JSON y se rompe fácil). Se previene: commits pequeños y un notebook por persona.

`````python
# Abre el archivo con los marcadores en VS Code
code patronus.txt
# -> decide qué versión queda (vale mezclar: "PATRONUS: ciervo y cierva")
# -> BORRA las 3 líneas de marcadores (<<<<<<<, =======, >>>>>>>) y guarda

# El ritual: add + commit
git add patronus.txt
git commit -m "fix/conflicto_patronus_resuelto"

# Comprueba que está todo limpio y que hay un commit de merge
git status
git log --oneline

# Limpieza: borra las ramas ya fusionadas y lista las tuyas (la * marca dónde estás)
git branch -d harry
git branch -d snape
git branch
`````

## 2.3 Entorno Python con `uv`: la receta viaja, la cocina no

El bug número 1 de los proyectos en equipo no es de código: es **"a mí me funciona"**. Pasa
cuando cada uno tiene un Python y unas librerías distintos. Solución: cada proyecto lleva su
propia **burbuja** (entorno virtual, la carpeta `.venv`) y su **receta** (`requirements.txt`).

- La **cocina** (`.venv`, miles de archivos) NUNCA se sube a Git.
- La **receta** (`requirements.txt`, unas líneas de texto) SÍ: con ella cualquiera rehace
  tu entorno exacto con 3 comandos.

Hoy la burbuja la crea **`uv`**: una herramienta modernísima y rapidísima para gestionar
Python y entornos. (Plan B universal si uv no arranca: `python -m venv .venv` + `pip install`.)

## Ronda 2 — El conflicto, a propósito

El TA nombra a **la dupla valiente** (2 personas). El resto hace una ronda suave:
rama `alumno/<tu-nombre>-v2`, retocar una línea de TU ficha, PR, revisión, merge (+1 punto).

**Guion exacto de la dupla** (coordinados por el chat del taller, no por vuestra cuenta):

1. Los dos: `git switch main && git pull`. Cada uno crea su rama `feature/lema-<nombre>`.
2. Los dos editáis **LA MISMA línea** de `lema.md` (la del `LEMA OFICIAL:`) con vuestro lema.
3. El TA da turno: **el primero** hace push + PR + merge YA.
4. **El segundo** hace push y abre su PR → **semáforo rojo: "This branch has conflicts that
   must be resolved"**. No has roto nada: Git ha protegido el trabajo del primero.
5. El segundo resuelve EN LOCAL (comandos abajo). Merge + delete branch.

`````python
# ===== PASOS DE LA DUPLA DE CONFLICTO =====

# Los dos: partir de main al día y crear vuestra rama
git switch main && git pull
git switch -c feature/lema-<tu-nombre>

# Los dos: editar LA MISMA línea de lema.md (la del LEMA OFICIAL:)
code lema.md
# -> cambia "LEMA OFICIAL: Datos limpios, modelito feliz." por TU lema y guarda

# Los dos: commit de vuestro lema
git add lema.md
git commit -m "docs/lema_<tu-nombre>"

# El PRIMERO (cuando el TA dé turno): push + PR + squash merge en la web
git push -u origin feature/lema-<nombre-primero>

# El SEGUNDO: push + PR -> semáforo rojo de conflicto en la web
git push -u origin feature/lema-<nombre-segundo>
`````

`````python
# ===== EL SEGUNDO RESUELVE EN LOCAL =====

# Trae main a tu rama: el conflicto aparece en TU archivo
git pull origin main

# Mira los marcadores del conflicto
cat lema.md
# <<<<<<< HEAD
# LEMA OFICIAL: el lema del compañero que mergeó antes
# =======
# LEMA OFICIAL: tu lema
# >>>>>>> feature/lema-tu-nombre

# Abre el archivo, decide el lema final (vale mezclar: "...Y café.")
# y BORRA las 3 líneas de marcadores
code lema.md

# El ritual: add, commit, push -> el PR se pone en verde SOLO
git add lema.md
git commit -m "fix/conflicto_lema_resuelto"
git push

# En la web: Squash merge + Delete branch. +3 puntos a la dupla.
`````

### Tres formas de resolver un conflicto (hoy practicamos dos)

1. **Git Bash** (el ritual de siempre): abrir el archivo, decidir, borrar marcadores,
   `add`, `commit`, `push`.
2. **VS Code + GitLens**: con el conflicto activo, VS Code pinta colores y botones
   *Accept Current Change* / *Accept Incoming* / *Accept Both* / *Compare Changes*, y el
   panel de Source Control lista los archivos en conflicto (*Merge Changes*). Clic, guardar,
   y los mismos `add` + `commit` + `push`.
3. **La web**: el botón *Resolve conflicts* del PR abre un editor en el navegador.
   Vale para un apuro; en proyectos reales se resuelve en local.

## 4.1 `git stash`: el bolsillo

Escenario: estás a mitad de un cambio (sin commitear) y necesitas cambiar de rama YA.

## 4.2 `fetch` vs `pull` (y quién es `origin/main`)

- `git fetch` = "ve a GitHub y CUÉNTAME qué hay de nuevo": descarga la información pero
  **no toca tus archivos**. Es espiar sin riesgo.
- `git pull` = `fetch` + `merge`: trae **y** mezcla en tu rama actual.

**¿Y ese `origin/main` que aparece por ahí?** Es la **foto de la última conversación con el
servidor**: Git recuerda dónde estaba `main` en GitHub la última vez que hablaste con él.
Son las "ramas remotas de seguimiento" y se listan aparte.

`````python
# Lista las ramas remotas que conoce tu Git (origin/main, origin/docs/...)
git branch -r

# Actualiza las "fotos del servidor" SIN tocar tus archivos
git fetch

# Vuelve a mirar: origin/main puede haber avanzado (si alguien pusheó algo)
git branch -r

# Puedes inspeccionar el remoto antes de mezclar:
git log origin/main --oneline

# Y cuando decidas mezclar, ya sí:
git pull
`````

## 4.3 `revert` vs `reset` vs `rebase`: los tres "deshacer"

| Quiero... | Comando | Cuándo |
|---|---|---|
| Deshacer el último commit LOCAL guardando cambios | `git reset HEAD~1` | Aún NO has hecho push |
| Deshacer un commit YA subido (compartido) | `git revert <hash>` | Crea un commit nuevo que invierte el viejo. NUNCA reescribas historia compartida |
| Reorganizar/reubicar commits sobre otra base | `git rebase ...` | Herramienta avanzada: hoy solo la nombramos. Prohibida en ramas compartidas |

**Regla simple: commit sin pushear → `reset`. Commit ya pusheado → `revert`.**

Variantes de `reset`: `HEAD~1` = una foto antes; `HEAD^` = lo mismo; `HEAD~2` = dos antes;
también vale el hash: `git reset a1b2c3d`. Modos: `--soft` (cambios al carrito),
`--mixed` (por defecto: cambios sueltos), `--hard` (borra commit Y cambios: peligroso).

**Bonus:** para corregir solo el MENSAJE del último commit (sin pushear):
`git commit --amend -m "mensaje/nuevo"`.

**Y el comodín universal:** `git reflog` lista TODOS tus movimientos, incluso commits
"perdidos" tras un reset. Git casi nunca pierde nada; lo único irrecuperable es lo que
nunca llegaste a commitear.

`````python
# Experimentos seguros en una carpeta de pruebas
cd ~/taller-github
mkdir laboratorio && cd laboratorio
git init
echo "v1" > notas.txt && git add notas.txt && git commit -m "add/notas_v1"
echo "v2" > notas.txt && git add notas.txt && git commit -m "update/notas_v2"
git log --oneline

# reset: borra la última foto, cambios sueltos en la carpeta
git reset HEAD~1
git log --oneline
git status

# revert: NO borra nada; crea un commit nuevo que invierte uno viejo
git revert <hash-de-tu-primer-commit>
git log --oneline
cat notas.txt

# reflog: la caja negra. TODO lo que hizo tu HEAD, incluso lo "borrado"
git reflog

# amend: corrige el mensaje del último commit (solo sin pushear)
git commit --amend -m "update/mensaje_corregido"
git log --oneline
`````

# Bloque 5 — EXTENSIÓN: dinámica "Owner + Forks" 

Aquí practicáis el flujo para contribuir a un proyecto **del que NO sois colaboradores**:
**fork → PR al repo original**. Es el flujo del open source... y el que usaréis para aportar
a cualquier repo público del mundo.

**Equipos de 3:** 1 **OWNER** (crea el repo) + 2 **CONTRIBUTORS** (forkean y contribuyen).

Aparece una rama nueva: **`develop`** (rama de integración del flujo *git-flow simplificado*).
En vuestro proyecto de grupo seguiréis usando SOLO `main`; esto es para que conozcáis el
otro flujo (lectura: nvie.com, en Recursos).

## 5.1 El OWNER crea el mini-proyecto DS

`````python
# ===== ROL: OWNER =====

# Crea la carpeta del proyecto y el repo local
cd ~/taller-github
mkdir proyecto-equipo && cd proyecto-equipo
git init

# Estructura estándar de un proyecto de datos
mkdir -p data notebooks src
touch README.md .gitignore data/diccionario.md src/.gitkeep notebooks/.gitkeep
`````

Contenido mínimo (ábrelo con `code .`):

- `README.md`: título del proyecto + descripción de 2 líneas + sección "Equipo" con los 3 nombres.
- `.gitignore`: `data/*.csv`, `.venv/`, `__pycache__/`, `.env`
- `data/diccionario.md`: qué columnas tendría vuestro dataset (nombre, tipo, descripción).

`````python
# ===== ROL: OWNER (continuación) =====

# Primer commit con TODO
git add -A
git commit -m "add/estructura_proyecto_equipo"

# En la web: New repository -> proyecto-equipo-<nombre-owner> -> público -> NO inicializar
# Y el ritual de conectar:
git remote add origin https://github.com/<owner>/proyecto-equipo-<nombre-owner>.git
git branch -M main
git push -u origin main

# Novedad del flujo git-flow: la rama de integración develop
git switch -c develop
git push -u origin develop
`````

## 5.2 Los CONTRIBUTORS: fork → clone → upstream → rama → push

Cada contributor, **en la web** del repo del owner: botón **Fork** (arriba a la derecha) →
*Create fork* en SU cuenta. El fork es una copia completa del repo que vive en el SERVIDOR,
en tu cuenta, y está VINCULADO al original (pero no se sincroniza solo).

`````python
# ===== ROL: CONTRIBUTOR =====

# Clona TU fork (ojo: la URL es la de TU cuenta)
cd ~/taller-github
git clone https://github.com/<TU-usuario>/proyecto-equipo-<nombre-owner>.git
cd proyecto-equipo-<nombre-owner>

# Comprueba: origin apunta a TU fork
git remote -v

# Añade el repo ORIGINAL como segundo remoto, por convención llamado "upstream"
git remote add upstream https://github.com/<owner>/proyecto-equipo-<nombre-owner>.git

# Ahora tienes dos remotos: origin (tu fork) y upstream (el original)
git remote -v
`````

`````python
# ===== ROL: CONTRIBUTOR (continuación): a trabajar =====

# Colócate en develop y actualízala desde el ORIGINAL (upstream)
git switch develop
git pull upstream develop

# Tu rama de tarea, como siempre
git switch -c feature/<tarea>-<tu-nombre>

# Haz tu tarea. Ideas:
#   notebooks/eda_<tu-nombre>.md   -> mini análisis exploratorio en markdown
#   src/limpieza_<tu-nombre>.py    -> una función de limpieza con su docstring
#   data/diccionario.md            -> ampliar el diccionario con 2 columnas nuevas
code .

# add + commit + push... pero el push va a TU fork (origin)
git add .
git commit -m "add/<tu_tarea>"
git push -u origin feature/<tarea>-<tu-nombre>
`````

### El PR al repo ORIGINAL

En la web de **TU fork**: botón **Contribute → Open pull request**. Ojo a los combos:

- **base**: `develop` del repo del OWNER (el original)
- **compare**: tu rama `feature/<tarea>-<tu-nombre>`

Crea el PR y pide revisión al owner. El owner: *Files changed* → *Start a review* →
comentario o *Request changes* → cuando esté bien, *Approve* → **Squash merge** →
*Delete branch*.

`````python
# ===== ROL: CONTRIBUTOR (tras el merge del owner): sincroniza tu fork =====

# Tu fork NO se sincroniza solo. Dos caminos equivalentes:

# Camino terminal:
git switch develop
git pull upstream develop      # actualiza tu develop local desde el ORIGINAL
git push origin develop        # y lo sube a TU fork

# Camino web: botón "Sync fork" en la página de tu fork.

# Limpieza de la rama de trabajo ya fusionada (local y remota)
git branch -d feature/<tarea>-<tu-nombre>
git push origin --delete feature/<tarea>-<tu-nombre>
`````

## 5.3 Cierre de la dinámica

Con todas las tareas fusionadas en `develop`, el OWNER hace el merge final a `main`
**por PR, nunca directo**: PR `develop → main` → *Squash merge* → y, opcional, etiqueta
de versión. Todos los contributors sincronizan sus forks una última vez.

`````python
# ===== ROL: OWNER (cierre) =====

git switch main
git pull

# En la web: New pull request con base main y compare develop -> Squash merge

# Etiqueta de versión (opcional, pero bonito)
git tag v1.0
git push origin v1.0

# ===== TODOS: última sincronización de forks =====
git switch develop
git pull upstream develop
git push origin develop
`````

## Qué aprendéis aquí que no se aprende con collaborators

- **fork ≠ clone**: el fork vive en el SERVIDOR (en tu cuenta); el clone, en tu portátil.
- **Dos remotos a la vez**: `origin` (tu fork) y `upstream` (el original).
- **Tu fork NO se sincroniza solo**: o *Sync fork* (web), o `pull upstream` + `push origin`.
- Se puede contribuir a **CUALQUIER repo público del mundo** sin tener permisos: fork → PR.

**Si se manda como deberes**, en el kickoff se comprobará:

- [ ] Repo del owner con `develop` fusionado en `main` (por PR).
- [ ] Los 2 PRs de los contributors, cada uno con al menos 1 revisión real.
- [ ] Los forks sincronizados (sin cartel "This branch is X commits behind").

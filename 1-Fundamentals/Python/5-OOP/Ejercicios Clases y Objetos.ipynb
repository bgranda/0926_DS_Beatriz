```python
class Tienda:

    def __init__(self, nombre, direccion, numero_empleados, ventas_ultimos_3meses, tipo = "Electrodomésticos", abierta = True):
        self.nombre = nombre
        self.direccion = direccion
        self.empleados = numero_empleados
        self.ventas = ventas_ultimos_3meses
        self.tipo = tipo
        self.abierta = abierta

    def características(self):
        return "Tienda: " + self.nombre + " Dirección: " + self.direccion + " Nº empleados: " + str(self.empleados) + " Ventas último trimestre por mes: " + str(self.ventas) + " Tipo de tienda: " + self.tipo + " Abierta: " + str(self.abierta)

    def total_ventas(self):
        return sum(self.ventas)

    def media_ventas_empleado(self):
        return sum(self.ventas) / len(self.ventas) / self.empleados

    def buscar_tienda(self):
        print(f"Nombre: {self.nombre}, Dirección: {self.direccion}")

    def ventas_ultimo_mes(self):
        return self.ventas[-1]

    def proyección_ventas(self, inversion):
        if inversion < 1000:
            for i in range(len(self.ventas)):
                self.ventas[i] = self.ventas[i] * 1.2
        elif inversion >= 1000:
            for i in range(len(self.ventas)):
                self.ventas[i] = self.ventas[i] * 1.5
        return self.ventas

```

```python
tienda1 = Tienda("Balay", "Paseo de la Castellana 151", 100, [4000, 5000, 3000])
tienda2 = Tienda("Bosch", "Avenida del Mediterráneo 320", 150, [3500, 6000, 7000])
tienda3 = Tienda("Siemens", "Calle de Rosalía de Castro 20", 75, [4250,7500, 8000])
```

```python
print(tienda1.características())
print(tienda1.total_ventas())
print(tienda1.media_ventas_empleado())
print(tienda1.buscar_tienda())
print(tienda1.ventas_ultimo_mes())
print(tienda1.proyección_ventas(500))
```

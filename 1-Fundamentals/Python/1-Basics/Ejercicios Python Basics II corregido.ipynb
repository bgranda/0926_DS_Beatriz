## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

````python
import math
grados = float(input("introduce el valor de los grados: "))
radianes = grados * (math.pi / 180)
print(f"El valor de {grados} º en radianes es: {radianes} rad")
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python
base = float(input("¿Cuál es el valor de la base del paralelogramo en cm: ?"))
altura = float(input("¿Cuál es el valor de la altura del paralelogramo en cm: ?"))
area = base * altura
print(f"El área del paralelogramo es: {area} cm^2")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10

````python
valor1 = float(input("introduce un valor: "))
valor2 = float(input("introduce el segundo valor: "))
print(f"El valor1 es: {valor1} y el valor2 es: {valor2}")
suma = valor1 + valor2
print(f"La suma es: {suma}")
print(f"¿La suma es igual a 10? {suma == 10}")
print(f"¿La suma es mayor que 10? {suma > 10}")
print(f"¿La suma es menor que 10? {suma < 10}")
````

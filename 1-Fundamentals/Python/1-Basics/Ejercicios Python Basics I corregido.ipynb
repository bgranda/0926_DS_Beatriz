## Ejercicio 5
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = (C × 9 / 5) + 32 

```python
temperatura_Celsius = input("Pon el valor de la temperatura en grados Celsius: ")
temperatura_Farenheit = float(temperatura_Celsius) * 9/5 + 32
print(f"El equivalente de {temperatura_Celsius} ºC en grados Farenheit es: {temperatura_Farenheit} F")
```

```python
# pasar un str a un numero entero, tienes que hacerlo en varios pasos
int("3.9")
```

```python
int(float("3.9")) # se va a quedar solo con la parte entera, no va a redondear
```

```python
int(float("3,9".replace(",","."))) # se tiene que cambiar primero la coma a un punto, para que lo entienda Python como número. Después lo pasa a float y después a entero
```

```python
barco1 = [(0,1), (1, 1)] # está en vertical, porque la columna siempre es 1 y cambia la fila
def coloca_barco(tablero, barco):
    for pieza in barco:
        tablero[pieza] = "0" # vete a tablero, vete a pieza y ponme una "O"
    return tablero # si yo tengo una función, y tengo una colección. Y luego lo modifico y te lo devuelvo, tengo un tablero modificado. Me duelve un np.array

tablero = coloca_barco(tablero, barco1)
print(tablero)
```

```python
# voy a disparar
def recibir_disparo(tablero, coordenada): # tengo un np.array que es el tablero, y una coordenada
    if tablero[coordenada] == "O": # si hay una "O", Tocado, has acertado en el barco
        tablero[coordenada] = "X"
        print("Tocado")

    elif tablero[coordenada] == "X" or tablero[coordenada] == "a": # puede haber otra opción de le has dado y sigue habiendo agua
        print("Agonia, deja de perder el tiempo, dispara en otro sitio")

    else:
        tablero[coordenada] = "a"
        print("Agua")

recibir_disparo(tablero, (5, 6))
print(tablero)
```

```python
def colaca_barco_plus(tablero, barco):
     # estoy haciendo esto para saber si pongo el barco dentro o fuera del tablero
     # nos devuelve el tablero si puede colocar el barco, si no devuelve False, y avise por pantalla (Esto es para la máquina)
    tablero_temp = tablero.copy() # es una copia, no es el original (el cual se mantiene sin modificar)
    num_max_filas = tablero.shape[0] # el tipo de objeto de shape es una tupla; tablero es un numpy array
    num_max_columnas = tablero.shape[1]
    for pieza in barco: # barco es una lista de tuplas, que tiene las coordenadas de donde se encuentra el barco. Pieza es una tupla
        fila = pieza[0] # pieza[0] es la primera tupla del barco
        columna = pieza[1] 

        if fila < 0 or fila >= num_max_filas: # si fila es menor que 0 o mayor o igual a 10 (recuerda que se nombran de 0 a 9)
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if columna < 0 or columna >= num_max_columnas: # lo mismo que pasa con las filas
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if tablero[pieza] == "O": # aquí hay otro barco, se quita la condición de "X", porque estoy colocando los barcos, no puede haber "X"
            print(f"No se puede poner la {pieza} porque hay otro barco")
            return False
        tablero_temp[pieza] = "O" # lo que he estado modificando es el temporal, no el original, para mantener modificado el tablero
    return tablero_temp # me devuelve mi tablero temporal, con el barco colocado completamente

        # tablero (10), es un tablero 10x10. Tiene 2 dimensiones, con 10 filas y 10 columnas, y tiene 100 elementos
```

```python
def crea_barco_aletorio(tablero, eslora=4, num_intentos=100):
    num_max_filas = tablero.shape[0] # el tipo de objeto de shape es una tupla; tablero es un numpy array (Tienes 10 filas)
    num_max_columnas = tablero.shape[1] # (Tienes 10 columnas) Al haber en función del tablero, me vale para cualquier tablero sin tener que definir la función de nuevo
    while True:
        barco = [] # tengo que construir mi barco
        # Construimos mi barco hipotético
        pieza_original = (random.randint(0, num_max_filas -1), random.randint(0, num_max_columnas -1))

        # (2,3) mi primera coordenada posible
        print(("pieza original:", pieza_original))
        barco.append(pieza_original) # [(2,3), (,), (,), (,)] tengo una eslora de 4, por lo que me faltan 3 coordenadas

        orientacion = random.choice(["N", "S", "E", "O"]) # se elige la orientación, ej: "S"
        print("Con orientacion:", orientacion)
        fila = pieza_original[0] # 2
        columna = pieza_original[1] # 3
        for i in range(eslora -1):
            if orientacion == "N":
                fila -= 1
            elif orientacion == "S":
                fila += 1
            elif orientacion == "E":
                columna += 1
            else:
                columna -= 1
            pieza = (fila, columna)
            barco.append(pieza) # [(2,3), (3,3), (4,3), (5,3)]
        tablero_temp = colaca_barco_plus (tablero, barco) # ahora llamo a coloca_barco_plus, para saber si cabe el barco o no hay ningún otro barco en esa posición
        if type(tablero_temp) == np.array:
            return tablero_temp # se para la función, en el primer return al que se llegue. Por eso, aquí se sale del bucle while 
        print("Tengo que intentar colocar otro barco")
```

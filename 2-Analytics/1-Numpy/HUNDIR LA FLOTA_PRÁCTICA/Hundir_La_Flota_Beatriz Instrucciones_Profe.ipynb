```python
# Videojuego de hundir la flota (se hace con Numpy y scripts)
import numpy as np
import random
```

```python
tablero = np.full(10, 10," ")
print(tablero)
```

```python
def crea_tablero(lado = 10):
    tablero = np.full((lado, lado), " ")
    return tablero
```

```python
tablero = crea_tablero(15)
print(tablero)
```

```python
barco1 = [(0,1), (1, 1)] # está en vertical, porque la columna siempre es 1 y cambia la fila
def coloca_barco(tablero, barco):
    for pieza in barco:
        tablero[pieza] = "0"
    return tablero # si yo tengo una función, y tengo una colección. Y luego lo modifico y te lo devuelvo, tengo un tablero modificado

tablero = coloca_barco(tablero, barco1)
print(tablero)
```

```python
# voy a disparar
def recibir_disparo(tablero, coordenada):
    if tablero[coordenada] == "0":
        tablero[coordenada] == "X"
        print("Tocado")

    elif tablero[coordenada] == "X" or tablero[coordenada] == "-": # puede haber otra opción de le has dado y sigue habiendo agua
        print("Agonia, deja de perder el tiempo, dispara en otro sitio")

    else:
        tablero[coordenada] = "-"
        print("Agua")

recibir_disparo(tablero, (5, 6))
print(tablero)
```

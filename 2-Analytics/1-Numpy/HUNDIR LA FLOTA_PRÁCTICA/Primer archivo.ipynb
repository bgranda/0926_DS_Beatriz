```python
# El jugador tiene que tener 2 tableros para jugar. Uno para colocar sus barcos y otro para señalar las posiciones en las que hace los disparos
# La máquina tiene que crear 2 tableros, crear sus barcos aleatoriamente y disparar aleatoriamente. 
# También debe tener dos tableros para señalar donde realiza los disparos
# Los tableros son siempre 10x10
```

```python
import numpy as np
import random
```

```python
tablero = np.full((10,10), " ")
tablero1_jugador = np.full((10,10), " ")
tablero2_jugador = np.full((10,10), " ")
tablero1_ordenador = np.full((10,10), " ")
tablero2_ordenador = np.full((10,10), " ")
```

```python
# El jugador primero, crea sus barcos y los coloca sus barcos en su tablero 1.
# Las entradas de las piezas se introducen: numero,numero
# 4 barcos: 1 barco de eslora 2, 1 barcos de eslora 3, 1 barco de eslora 4 y 1 barco de eslora 5.

def crea_barco(pieza):
    barco = []
    piezas_faltan = pieza

    while piezas_faltan > 0:
        entrada = input(f"Dame las coordenadas de tu pieza [x,y]:")
        x_str,y_str = entrada.split(",")
        x, y = int(x_str.strip()), int(y_str.strip())

        barco.append((x,y))
        piezas_faltan -= 1
    print(barco)

barco1 = crea_barco(2)
barco2 = crea_barco(3)
barco3 = crea_barco(4)
barco4 = crea_barco(5)
```

```python
def coloca_barco (tablero1_jugador, barco):
    tablero_temp = tablero1_jugador.copy()
    num_max_filas = tablero1_jugador.shape[0]
    num_max_columnas = tablero1_jugador.shape[1]
    for pieza in barco:
        fila = pieza[0]
        columna = pieza[1]
        if fila < 0 or fila >= num_max_filas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if columna < 0 or columna >= num_max_columnas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if tablero[pieza] == "O":
            print(f"No puedo poner la pieza {pieza} porque hay otro barco")
            return False
        tablero_temp[pieza] = "O"
    return tablero_temp
```

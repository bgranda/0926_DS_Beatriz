```python
def crea_barco(piezas):
    barco = []
    piezas_faltan = piezas

    while piezas_faltan > 0:
        try:
            entrada = input(f"Dame las coordendas de tu pieza [x,y]:")
            #Separar por coma y limpiar espacios
            partes = entrada.split(",")
            if len(partes) != 2:
                print("Formato inválido. Usa formato x,y.")
                continue

            x_str, y_str = partes
            x = int(x_str.strip())
            y = int(y_str.strip())

            barco.append((x,y))
            piezas_faltan -=1

        except ValueError:
            print("Coordenadas inválidas. Deben ser números enteros")
    print(barco)
    yield barco
```

```python
def coloca_barco (tablero1_US, barco):
    for pieza in barco:
        tablero1_US[pieza] = "O"
    return tablero1_US
```

```python
def colaca_barco_plus(tablero1_Japon, barco):
    tablero_temp = tablero1_Japon.copy()
    num_max_filas = tablero1_Japon.shape[0]
    num_max_columnas = tablero1_Japon.shape[1]
    for pieza in barco: 
        fila = pieza[0]
        columna = pieza[1] 

        if fila < 0 or fila >= num_max_filas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if columna < 0 or columna >= num_max_columnas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if tablero[pieza] == "O" or tablero[pieza] == "X":
            print(f"No se puede poner la {pieza} porque hay otro barco")
            return False
        tablero_temp[pieza] = "O"
    return tablero_temp
```

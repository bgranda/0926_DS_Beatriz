```python
import numpy as np
import random
```

```python
tablero = np.full((10,10), " ")
print(tablero)
```

```python
def crea_barco(tablero, fila_ini, columna_ini, eslora, orientacion):
    """
    Genera las coordenadas de un barco según la orientación dada.
    No modifica el tablero.
    
    tablero: numpy.ndarray (matriz del tablero con strings)
    fila_ini: int (fila inicial)
    columna_ini: int (columna inicial)
    eslora: int (longitud del barco)
    orientacion: str ('N', 'S', 'E', 'O')
    """
    num_max_filas, num_max_columnas = tablero.shape

    # Validación de coordenadas iniciales
    if not (0 <= fila_ini < num_max_filas):
        raise ValueError("La coordenada inicial de fila está fuera del tablero")
    if not (0 <= columna_ini < num_max_columnas):
        raise ValueError("La coordenada inicial de columna está fuera del tablero")

    orientacion = orientacion.strip().upper()
    if orientacion not in ("N", "S", "E", "O"):
        raise ValueError("Orientación inválida. Debe ser N, S, E u O.")

    # Validar que el barco cabe en el tablero según orientación
    if orientacion == "N" and fila_ini - (eslora - 1) < 0:
        raise ValueError("El barco no cabe hacia el norte.")
    elif orientacion == "S" and fila_ini + (eslora - 1) >= num_max_filas:
        raise ValueError("El barco no cabe hacia el sur.")
    elif orientacion == "E" and columna_ini + (eslora - 1) >= num_max_columnas:
        raise ValueError("El barco no cabe hacia el este.")
    elif orientacion == "O" and columna_ini - (eslora - 1) < 0:
        raise ValueError("El barco no cabe hacia el oeste.")

    # Construir el barco
    barco = [(fila_ini, columna_ini)]
    fila, columna = fila_ini, columna_ini
    for _ in range(eslora - 1):
        if orientacion == "N":
            fila -= 1
        elif orientacion == "S":
            fila += 1
        elif orientacion == "E":
            columna += 1
        elif orientacion == "O":
            columna -= 1
        barco.append((fila, columna))

    return barco
```

```python
def coloca_barco(tablero, barco):
    """
    Coloca el barco en el tablero marcando las posiciones con 'O'.
    Verifica colisiones directas.
    """
    for pieza in barco:
        if tablero[pieza] != " ":
            raise ValueError(f"Colisión detectada en {pieza}. Ya hay un barco allí.")
        tablero[pieza] = "O"
    return tablero
```

```python
if __name__ == "__main__":
    tablero = np.full((10, 10), " ")  # Tablero vacío

    try:
        # Crear barco sin colocarlo
        barco_coords = crea_barco(tablero, 0, 1, 2, 'S')
        print("Coordenadas generadas:", barco_coords)

    # Colocar barco en el tablero
        tablero = coloca_barco(tablero, barco_coords)
        print("\nTablero con barco:")
        print(tablero)
    except ValueError as e:
        print("Error:", e)
```

```python
def crea_tablero(lado = 10):
    tablero = np.full((lado, lado), " ")
    return tablero
```

```python
# voy a disparar
def recibir_disparo(tablero, coordenada):
    if tablero[coordenada] == "0":
        tablero[coordenada] == "X"
        print("Tocado")

    elif tablero[coordenada] == "X":
        print("Agonia, deja de perder el tiempo, dispara en otro sitio")

    else:
        tablero[coordenada] = "-"
        print("Agua")

recibir_disparo(tablero, (5, 6))
print(tablero)
```
